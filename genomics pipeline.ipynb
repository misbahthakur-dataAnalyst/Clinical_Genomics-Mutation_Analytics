```python
import pandas as pd
import numpy as np

#messy clinical genomics dataset 
practice_data={
    "Sample_ID":[1001,1002,1003,1004,1005,1006],
    "Researcher_Name":["misbh","ananya","zaid","adeeba","kabir","misbah"],
    "Lab_Zone":["Genetics core",np.nan,"Bio-informatics","Genetics core",np.nan,"Bio-informatics"],
    "DNA_Mutation_count":[12,np.nan,8,24,np.nan,14],
    "Patient_Weight_Kg":[65,72,np.nan,88,54,np.nan],
    "Sequence_Verified":["yes","yes","no",np.nan,"yes","no"]
}

df_practice= pd.DataFrame(practice_data)
df_practice.to_csv("raw_mutation_data.csv",index=False)

print("success!'raw_mutation_data.csv'generated.")
print("___raw uncleaned data preview___")
print(df_practice)
```

```python
df_clean= df_practice.drop_duplicates()
print(df_clean)
```

```python
df_clean["Lab_Zone"]= df_clean["Lab_Zone"].fillna("General Zone")
df_clean=df_clean.bfill().ffill()

print("___missing data successfully resolved___")
print(df_clean)
```

```python
# lambda expression evaluation framework tracking mutations
# condition: if mutation count >=15 mapping to"High Priority",else "Standard"

df_clean["Alert_Status"]= df_clean["DNA_Mutation_count"].apply(lambda x: "High Priority"if x>= 15 else "Standard")

df_clean.to_csv("cleaned_genomics_analytics.csv",index=False)

print("___ PREMIUM CLINICAL GENOMICS DASHBOARD___")
print(df_clean)
print("\n SUCCESS! Cleaned file 'cleaned_genomics_analytics.csv' is saved and project is complete.")
```
